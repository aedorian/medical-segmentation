```python
import sys
import os
import json
import shutil
import tempfile
import time
import glob
from urllib.parse import parse_qs

import matplotlib.pyplot as plt
import numpy as np
import nibabel as nib
```

```python
data_dir = "/datasets/tdt4265/mic/open/HNTS-MRG"

shapes = []
min_intensities = []
max_intensities = []
tumor_one_only = 0
tumor_two_only = 0
tumor_both = 0

i = 0

part = "/test" # "/train"

patients = os.listdir(data_dir + part)
n = len(patients)
print(f"Anayzing {part} training data... ({n} patients)")
for p in patients:

    img_path = f"{data_dir}/{part}/{p}/preRT/{p}_preRT_T2.nii.gz"
    mask_path = f"{data_dir}/{part}/{p}/preRT/{p}_preRT_mask.nii.gz"
    
    img = nib.load(img_path).get_fdata()
    mask = nib.load(mask_path).get_fdata()
    
    # Analyze classes
    classes = np.unique(mask)
    if 1 in classes and 2 in classes:
        tumor_both += 1
    elif 1 in classes and 2 not in classes:
        tumor_one_only += 1
    elif 2 in classes and 1 not in classes:
        tumor_two_only += 1
        
    # Add intensities of MRI scan
    min_intensity = np.min(img)
    max_intensity = np.max(img)
    min_intensities.append(min_intensity)
    max_intensities.append(max_intensity)
    
    # Check that image and mask shapes are equal
    print(f"({i+1}/{n}) image shape: {img.shape}, mask shape: {mask.shape}, classes: {classes}, min_intensity: {min_intensity}, max_intensity: {max_intensity}")
    
    # Add shape to different shapes
    if img.shape not in shapes:
        shapes.append(img.shape)
        
    i += 1
    
print(f"Number of patients with tumor 1 only: {tumor_one_only}")
print(f"Number of patients with tumor 2 only: {tumor_two_only}")
print(f"Number of patients with both tumors: {tumor_both}")

print(f"-----\n{len(shapes)} different shapes found:")
print(shapes)

print(f"-----\nMaximum intensity: {np.max(max_intensities)} (average: {np.average(max_intensities)})")
print(f"Minimum intensity: {np.max(min_intensities)} (average: {np.average(min_intensities)})")
```
